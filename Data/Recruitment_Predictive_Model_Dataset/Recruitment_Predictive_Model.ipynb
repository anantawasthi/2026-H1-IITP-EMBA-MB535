```python
# Exploratory Data Analysis (EDA) - Univariate data analysis
data.describe()
```

```python
categorical_cols = data.select_dtypes(include=['object']).columns
categorical_cols
```

```python
# create a frequnency table for each categorical variable alogn with a bar plot. do not include candidate id in the analysis
for col in categorical_cols:
    if col != 'Candidate_ID':
        display(f"Frequency table for {col}:")
        display(data[col].value_counts())
        plt.figure(figsize=(8, 4))
        sns.countplot(x=col, data=data)
        plt.title(f"Distribution of {col}")
        plt.xticks(rotation=45)
        plt.show()
```

```python
# create boxplots for each numerical variable to identify outliers
numerical_cols = data.select_dtypes(include=[np.number]).columns
```

```python
# create box plots for each numerical variable to identify outliers
for col in numerical_cols:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=data[col])
    plt.title(f"Box plot of {col}")
    plt.show()
```
